```python
import calc as cl
```

```python
op = int(input("Digite a operação (1 - soma, 2 - subtração, 3 - multiplicação, 4 - divisão, 5 - potência, 6 - radiciação): "))

if op == 1:
    num1 = int(input("Digite o primeiro número: "))
    num2 = int(input("Digite o segundo número: "))
    print(cl.soma(num1, num2))

elif op == 2:
    num1 = int(input("Digite o primeiro número: "))
    num2 = int(input("Digite o segundo número: "))
    print(cl.subtracao(num1, num2))

elif op == 3:
    num1 = int(input("Digite o primeiro número: "))
    num2 = int(input("Digite o segundo número: "))
    print(cl.multiplicacao(num1, num2))

elif op == 4:
    num1 = int(input("Digite o primeiro número: "))
    num2 = int(input("Digite o segundo número: "))
    print(cl.divisao(num1, num2))

elif op == 5:
    num1 = int(input("Digite a base: "))
    num2 = int(input("Digite o expoente: "))
    print(cl.potencia(num1, num2))

elif op == 6:
    num1 = int(input("Digite o número: "))
    print(cl.raiz_quadrada(num1))
```
